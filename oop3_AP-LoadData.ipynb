# Example: reading data in a class

### Simple data loader Class


````python
class DataModel:
    def __init__(self, filename: str):
        self.filename = filename
        self.data: list | None = None
        self.header: str | None = None

    def load_data(self) -> tuple[str, list]:
        data = []
        with open(self.filename, "r") as f:
            header =f.readline().strip()  # Skip header line
            for line in f:
                data.append(float(line))
            self.data = data
            self.header = header
        print(f"Loaded {len(data)} rows from {self.filename}")
        return header, data


data1_model = DataModel("files/data1.csv")
header, data1 = data1_model.load_data()
````

````python
class DataModel2:
    def __init__(self, filename: str, fs: int):
        self.filename = filename
        self.fs = fs
        self.data: list | None = None
        self.header: str | None = None
        self.label = f"ECG @ {fs:.0f} Hz"

    def load_data(self):
        data: list[float] = []
        with open(self.filename, "r") as f:
            header =f.readline().strip()  # Skip header line
            for line in f:
                parts = line.split(",")
                data.append(float(parts[1]))
            self.data = data
            self.header = header
        print(f"Loaded {len(data)} rows from {self.filename}")
        self.header = header
        data = data
        return data
````

````python

dataa_samplerate = 300
datab_samplerate = 150
model_a = DataModel2("files/ECG_300Hz.csv", fs =300)
dataa = model_a.load_data()
model_b = DataModel2("files/ECG_150Hz.csv", fs=150)
datab = model_b.load_data()
labelb = f"ECG @ {datab_samplerate} Hz"

print(f"{model_a.label:15}: {len(dataa)=}, og {dataa_samplerate=}Hz")
print(f"{model_b.label:15}: {len(datab)=}, og {datab_samplerate=}Hz")
````

````python

plt.plot(dataa, label=model_a.label)
plt.plot(datab, label=model_b.label)
plt.title("ECG data")
plt.legend()
plt.xlabel("Time [s]")
plt.show()
````

Dette er en fin løsning til at plotte meget simpelt data.

Hvis man rammer ind i at skal gøre dette mange gange i samme program:
- Svært at overskue
- Svært at fejlfinde
- Meget repetitiv kode

````python
class Signal:
    def __init__(self, model: DataModel2):
        # Attributes
        data = model.load_data()
        time = [i/model.fs for i in range(len(data))]
        self.data = data
        self.fs = model.fs
        self.time = time
        self.label = model.label

    # methods
    def plot(self, ax=None):
        """Method to plot data"""
        if not ax:
            ax = plt.subplot(1, 1, 1)
        ax.plot(self.time, self.data, label=self.label)
        return ax


# Normal Function
def finalize_figure():
    """Finalizes figure with visual queues"""
    plt.legend()
    plt.xlabel("time [s]")
    plt.ylabel("mV")
    plt.show()


# Definer 2 sigler ud fra dataa og datab
signal_a = Signal(model_a)
signal_b = Signal(model_b)

# Plot figurerne oven i hinanden, ved at bruge en figurehandle
ax = plt.subplot(1, 1, 1)
signal_a.plot(ax)
signal_b.plot(ax)
finalize_figure()
````

Herunder er der noget pseudokode, som i allerede har lavet i C på jeres ESP32.


### **Pseudocode: Peak Detection**

```
INITIALIZE:
    thr ← initial_threshold          # main threshold for peak detection
    record_value ← 0                       # current peak candidate value
    record_time ← 0                      # index of current peak candidate
    indecies ← []                       # list to store detected peak positions
    t ← 0                            # sample index

FOR each value in signal:
    IF value > thr DO
        # Above threshold: update peak candidate if higher than previous
        IF value > record_value DO
            record_value ← value
            record_time ← t
        ENDIF
        
    ELSE DO
        # Below threshold: check if we had a peak candidate AND signal dropped below (thr)
        IF record_value > 0 DO
            # Confirm peak and store its position
            ADD recordt TO indecies
            # Reset candidate
            record_value ← 0
            recordt ← 0
        ENDIF
        
    ENDIF
    t ← t + 1
    
END FOR

OUTPUT indecies
```

## Answer: Load Data, step 1-4
</div>

````python
import matplotlib.pyplot as plt

class PatientInfo:
    def __init__(self, name, dob, systolic_bp, diastolic_bp, ef_percent):
        self.name = str(name)
        self.gender = str(dob)
        self.height = float(systolic_bp)
        self.weight = float(diastolic_bp)
        self.bmi = float(ef_percent)

    def print_info(self):
        print(f"Name:{self.name:>20}, Gender:{self.gender:>6}, Height:{self.height:>4.2f}m, Weight:{self.weight: >2.0f}kg, BMI:{self.bmi:.0f}")
        

class MyDataLoader:
    def __init__(self, filename: str):
        self.filename = filename

    def load_data(self) -> tuple[list, list]:
        data = []
        with open(self.filename, "r") as f:
            headers =  f.readline().split(",") # Første linje er headers
            for row in f:
                parts = row.split(",")
                p = PatientInfo(*parts) # Dette ved i ikke, dette unpacker simpelthen et dataset
                data.append(p)

            self.headers = headers
            self.data = data
            print(f"Data loaded from {self.filename}")
        return headers, data


dataloader = MyDataLoader("files/exercise_people_small.csv")
headers, data = dataloader.load_data()

for p in data:
    p.print_info()
````

## Answer: Show data
</div>

````python

import matplotlib.pyplot as plt
        
class MyDataVizualizer:
    def __init__(self, data: list, headers: list):
        self.data: list[PatientInfo] = data
        self.headers = headers

    def plot_data(self):
        heights_male = [p.height for p in self.data if p.gender == "male"]
        weights_male = [p.weight for p in self.data if p.gender == "male"]
        heights_female = [p.height for p in self.data if p.gender != "male"]
        weights_female = [p.weight for p in self.data if p.gender != "male"]

        plt.scatter(heights_male, weights_male,label="Mænd")
        plt.scatter(heights_female, weights_female,label="Kvinder")
        plt.legend()
        plt.xlabel("height [m]")
        plt.ylabel("weights [kg]")
        plt.title(f"Sammenhænget mellem height og weight for mænd og kvinder")
        plt.show()

dataloader = MyDataLoader("files/exercise_people.csv")
header, data = dataloader.load_data()
my_viz = MyDataVizualizer(data,header)
my_viz.plot_data()
````
